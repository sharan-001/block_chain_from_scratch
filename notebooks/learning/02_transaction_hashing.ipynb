```python
# creating a function for this 
def hashcode(transaction3):
    json_tranasaction3 = json.dumps(transaction3)
    encoding = json_tranasaction3.encode()
    
    return hashlib.sha256(encoding).hexdigest()

tra = {
    "sender":"hello",
    "reciver":"ignore",
    "amount":10000000
}
p1 = hashcode(tra)
print(p1)
tra["amount"]=1010
hash2 = hashcode(tra)
print(hash2)
tra["hash"] = hashcode(tra)
print(tra)
# now we are going to real block chain 
print(tra["hash"])
print(hashcode(tra))
```
